## 4. Vector Spaces

A **Vector Space** is a set $V$ of elements called **vectors**, having operations of addition and scalar multiplication defined on it that satisfy the following conditions ($u$, $v$, and $w$ are arbitrary elements of $V$, and c and d are scalars.)

### Closure Axioms

1. The sum $u + v$ exists and is an element of $V$. ($V$ is closed under addition.)
2. $cu$ is an element of $V$. ($V$ is closed under scalar multiplication.)

### Addition Axioms

3. $u + v = v + u$ (commutative property)
4. $u + (v + w) = (u + v) + w$ (associative property)
5. There exists an element of $V$, called a **zero vector**, denoted $0$, such that $u+0 = u$
6. For every element $u$ of $V$, there exists an element called a **negative** of $u$, denoted $-u$, such that $u + (-u) = 0$.

### Scalar Multiplication Axioms

7. $c(u+v) = cu + cv$
8. $(c + d)u = cu + du$
9.  $c(du) = (cd)u$
10. $1u = u$

### Definition of a basis of a vector space

> A finite set of vectors ${v_1,\dots, v_n}$ is called a **basis** of a *vector space* $V$ if the set *spans* $V$ and is *linearly independent*. 
>i.e. each vector in $V$ can be expressed uniquely as a *linear combination* of the vectors in a basis.

Determine which vector space axioms hold and which do not.

- The circle with center at the origin and radius zero is the additive identity for U
- For any circle C in U, the circle -C with center at the origin and radius −r, where r is the radius of C, is the additive inverse of C, because C+(−C)=0.
- For any scalar k and any circle C in U, the circle kC with center at the origin and radius |k|r, where r is the radius of C, is also in U.
- For any scalar k and any circles C1 and C2 in U, k(C1+C2) = circle with center at the origin and radius k(r1+r2); kC1+kC2 = circle with center at the origin and radius kr1+kr2 where r1 and r2 are the radii of C1 and C2, respectively.

- For any circle C in U, 1C=C, because multiplying by a scalar of 1 does not change the circle. 
- So, all of the vector space axioms hold for the set U of all circles in R2 having center at the origin. Therefore, U is a vector space.

&#9989; **<font color=red>DO THIS:</font>** Let $v_1$ and $v_2$ span a vector space $V$. 
Let $v_3$ be any other vector in $V$. 
Show that $v_1$, $v_2$, and $v_3$ also span $V$.

Since v1 and v2 span V, any vector v in V can be expressed as a linear combination of v1 and v2:

v = av1 + bv2 where a and b are real numbers.

Now, let x, y, and z be real numbers such that:

v = xv1 + yv2 + zv3

We want to show that we can always find values of x, y, and z that satisfy this equation for any vector v in V.

Substituting the expression for v in terms of v1 and v2 into the above equation, we get:

xv1 + yv2 + zv3 = x(av1 + bv2) + y(av1 + bv2) + zv3

Simplifying, we get:

(xa + ya)v1 + (xb + yb)v2 + zv3 = v

Since v1 and v2 span V, we know that any vector in V can be expressed as a linear combination of v1 and v2. Therefore, we can rewrite the left-hand side of the above equation as:

xa + ya = p; xb + yb = q; where p and q are real numbers.

Substituting these values back into the equation, we get:

pv1 + qv2 + zv3 = v

Since v1, v2, and v3 are linearly independent, we know that v3 cannot be expressed as a linear combination of v1 and v2. Therefore, we can choose z such that:

pv1 + qv2 + zv3 = v

This shows that any vector v in V can be expressed as a linear combination of v1, v2, and v3. Therefore, v1, v2, and v3 span V.

&#9989; **<font color=red>DO THIS:</font>** Is the set of nonzero row vectors of any matrix in reduced row echelon form linearly independent? Discuss in your groups and include your thoughts below.

Yes. Since the row vectors are all in reduced row echelon form, their leading 1s are in different columns. Assume that row1 has the leading 1 in the first column. Then, the only way the first entry of the linear combination can be 0 is if c1 = 0.

Similarly, since row1 has the leading 1 in the first column, the only way the second entry of the linear combination can be 0 is if c2 = 0. Continuing in this way for each column, we see that all of the constants must be 0 for the linear combination to be equal to the zero vector.

Therefore, the set of nonzero row vectors of any matrix in reduced row echelon form is linearly independent.
